1. Створити змінні, що посилаються на два цілих числа, що однакові за значенням, де значення належить проміжку від -5 до 256. Перевірте, чи будуть ці змінні рівні тільки за значенням або ж ще будуть посилатися на один і той самий обʼєкт в памʼяті? Наведіть код та дайте текстову відповідь нижче.

11. Створіть список, де елементами цього списку також є списки. Отримай перший елемент з останнього рядка та виведи значення на екран.

# VI. Множини (Set).

17. Створити множину без елементів. Після цого за допомогою методу додайте кілька різних елементів до множини. Чи множини є змінним типом даних?

```python
my_set = set()
my_set.add("car")
my_set.add(1)

print("My set:", my_set)
```

Так, множини є змінними — можна додавати або видаляти елементи після створення.

18. Створити множину, маючи список my_list = [1, 1, 2, 67, 67, 8, 9]. Пояснити, чому "зникли" деякі елементи.

```python
my_list = [1, 1, 2, 67, 67, 8, 9]
my_set = set()
my_set.update(my_list)
print(my_set)
```

Множина зберігає лише унікальні елементи, тому дублікати були видалені.

19. Створіть дві множини. Продемонстуйте над ними операції: обʼєднання, різниці, пересічі та симетричної різниці. Використовуйте методи, що не змінюють множини, а створюють нові.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union_set = set1.union(set2)
print("Union:", union_set)

intersection_set = set1.intersection(set2)
print("Intersection:", intersection_set)

difference_set = set1.difference(set2)
print("Difference (set1 - set2):", difference_set)

symmetric_diff_set = set1.symmetric_difference(set2)
print("Symmetric Difference:", symmetric_diff_set)
```

# VII. Словники (Dictionary).

20. Створіть пустий словник. До нього додайте чотири пари елементів такі, щоб їхні ключі були різних типів. Чи може список бути ключем? Чому?

```python
empty_dict = {}

empty_dict["color"] = "red"
empty_dict["year"] = 1991
empty_dict["weight"] = 300.5
empty_dict[True] = "boolean_key"

print("Dictionary:", empty_dict)
```

Ні, список не може бути ключем, оскільки він є змінним типом даних. Ключі повинні бути незмінними.

21. Створіть словник, де значенням в одній з пар теж буде словник, який теж має вкладений словник. Виведіть на екран значення, що міститься в словнику, що знаходиться на найнижчому рівні ієрархії вкладеності (найбільш внутрішній).

```python
nested_dict = {
    "level1": {
        "level2": {
            "level3": "Deep Value"
        }
    }
}

deep_value = nested_dict["level1"]["level2"]["level3"]
print("Deep Value:", deep_value)
```
