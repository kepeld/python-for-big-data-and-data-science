5. (9б). Створіть клас 3D-фігура без атрибутів, з методом get_volume для отримання об'єму фігури, що повертає 0 та __add __, який приймає self та other в якості аргументів, а повертає суму об'ємів 3d-фігур self та other.

```python
class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        if isinstance(other, Figure3D):
            return self.get_volume() + other.get_volume()
        else:
            raise TypeError("Can only add Figure3D objects")

fig = Figure3D()
print("Volume of a basic 3D figure (should be 0):", fig.get_volume())
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        # Volume of cylinder: π * r^2 * height
        return math.pi * self.radius ** 2 * self.height

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        # Volume of cube: side^3
        return self.side ** 3

cyl = Cylinder(3, 5)
cube = Cube(4)

print("Volume of cylinder:", cyl.get_volume())
print("Volume of cube:", cube.get_volume())
print("Total volume (cylinder + cube) using __add__:", cyl + cube)
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, wallet_id, initial_balance=0):
        self.__id = wallet_id
        self.__balance = initial_balance

    @property
    def balance(self):
        return self.__balance

    @property
    def wallet_id(self):
        return self.__id

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance is {self.__balance}.")
        else:
            print("Amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance is {self.__balance}.")
        else:
            print("Insufficient funds or invalid amount.")

wallet = EWallet("wallet_001", 100)
print("Initial wallet balance:", wallet.balance)
wallet.deposit(50)
wallet.withdraw(30)
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __repr__(self):
        return f"{self.name}: ${self.price}"

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        if isinstance(item, Item):
            self.items.append(item)
            print(f"Added item: {item}")
        else:
            print("Only Item objects can be added.")

    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)
            print(f"Removed item: {item}")
        else:
            print("Item not in the cart.")

    def show_items(self):
        print("Shopping Cart Items:")
        for item in self.items:
            print(item)

cart = ShoppingCart()
item1 = Item("Laptop", 1200)
item2 = Item("Mouse", 25)
cart.add_item(item1)
cart.add_item(item2)
cart.show_items()
cart.remove_item(item1)
cart.show_items()
```

10. (13б). Створіть клас Vehicle з атрибутами brand (manufacturer) та production_age. Створіть ще один клас Engine з такими атрибутами, як type (ex. petrol) та size (volume, litres). Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age):
        self.brand = brand
        self.production_age = production_age

class Engine:
    def __init__(self, engine_type, size):
        self.engine_type = engine_type
        self.size = size

class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, engine_type, size):
        Vehicle.__init__(self, brand, production_age)
        Engine.__init__(self, engine_type, size)

    def __repr__(self):
        return (f"Car(brand={self.brand}, production_age={self.production_age},"
                f"engine_type={self.engine_type}, size={self.size})")

my_car = Car("Toyota", 5, "petrol", 2.0)
print("Car object:", my_car)
print("Method Resolution Order for Car:")
for cls in Car.mro():
    print(cls)
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekDayIterator:
    def __init__(self):
        self.days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration

week_iterator = WeekDayIterator()

for day in week_iterator:
    print(day)
```
